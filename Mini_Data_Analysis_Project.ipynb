```python
sales_data = [
    ["Ali", "Laptop", 80000],
    ["Sara", "Phone", 50000],
    ["Ahmed", "Tablet", 35000],
    ["Hina", "Laptop", 90000],
    ["Usman", "Phone", 45000]
]

total_sales = sum(sales for name, product, sales in sales_data)

avg_sales = total_sales / len(sales_data)

highest_salesperson = max(sales_data, key=lambda x: x[2])

lowest_salesperson = min(sales_data, key=lambda x: x[2])


print("\n================ Sales Analysis Report ================")

for name, product, sales in sales_data:

    if sales >= 80000:
        performance = "Excellent"
    elif sales >= 60000:
        performance = "Good"
    elif sales >= 40000:
        performance = "Average"
    else:
        performance = "Needs Improvement"

    print(name, ":", product, ":", sales, ":", performance)

print("--------------------------------------------------------")

print("Total Sales:", total_sales)
print("Average Sales:", avg_sales)
print("Highest Salesperson:", highest_salesperson)
print("Lowest Salesperson:", lowest_salesperson)

print("========================================================")
```
